```python
class Animal:
    def __init__(self,species,name,age,medication):
        self.species=species
        self.name=name
        self.age=age
        self.medications=medication
    
    def get_specie_description(self):
        print(f"This is a {self.specie}, it's name is {self.name} and it's age is {self.age}")
        if self.medication:
            print(f"The Recommended Medications are")
            for i,k in enumerate(self.medication.items()):
                print(f"The Medicine is {k[0]} and needed {k[1]}")
        else:
            print(f"it doesn't need a medication")

    def set_medications(self,medicine,dose,types):
        self.medications.append({
            types:dose,
            
        })
    
    def get_medications(self):

        for i,k in enumerate(self.medications):
            print(f"The medicine info is : \n")
            print(f"doses are {k['dose']} and info is {k['type']}")


```

```python
class Specie1(Animal):
    def __init__(self,specie,name,age,medication):
        super().__init__(specie,name,age,medication)
        self.age=0
        self.name=name 
        self.specie=specie
        self.medication=medication
    
    def make_sound(self):
        if "cat"in self.specie:
            print(" Meow!")
        elif "Dog" in self.specie:
            print("Woof!")
        elif "wolf" in self.specie:
            print("oooooo")
    
    
        
    
```

```python
class Shelter:
    def __init__(self, *args):
        self.animals = args
    
    def introduce_animal(self,**kwargs):

        for animal in self.animals:
            print()
            animal.make_sound()

            if kwargs.get("info"):
                animal.get_specie_description()
            
                
                
            

```
