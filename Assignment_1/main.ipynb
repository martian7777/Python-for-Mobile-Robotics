```python
class Animal:
    def __init__(self,species,name,age="unknown",medication=None):
        self.species=species
        self.name=name
        self.age=age
        self.medications= medication if medication is not None else {}
    
    def get_species_description(self):
        print(f"This is a {self.species}, it's name is {self.name} and it's age is {self.age}")
       

    def set_medications(self, medicine, dose):
       self.medications[medicine] = dose
    
    def get_medications(self):
     if self.medications:
        print("It needs the following medications:")

        for i, (medicine, dose) in enumerate(self.medications.items(), start=1):
            print(f"{i}. {medicine}: {dose}")
     else:
        print("It does not need medication.")
        


```

```python
class Dog(Animal):
    def __init__(self,name,age="unknown",medication=None):
        super().__init__("Dog",name,age,medication)
        
    
    def make_sound(self):
        return "Woof!"
    
    
        
    
```

```python
class Cat(Animal):
    def __init__(self,name,age="unknoww",medication=None):
        super().__init__("Cat",name,age,medication)
        
    
    def make_sound(self):
            print(" Meow!")
        
    
   
        
    
```

```python
class Shelter:
    def __init__(self, *args):
        self.animals = args
    
    def introduce_animal(self,**kwargs):

        for animal in self.animals:
            print()

            print(animal.make_sound())
            
            if kwargs.get("info"):
                animal.get_species_description()
                
            
            if kwargs.get("medication"):
                animal.get_medications()

            
                
                
            

```

```python

# Create medication plan
medicine = {
    "Antibiotic": "12 mg",
    "Penicillin": "100 mg"
}

# Create animals from different species
cat1 = Cat("Leo", 12)

dog1 = Dog("Amma", 5, medicine)

dog2 = Dog("Pluto")

# Create shelter
s = Shelter(cat1, dog1, dog2)
```

```python
s.introduce_animal(info=True)
```
