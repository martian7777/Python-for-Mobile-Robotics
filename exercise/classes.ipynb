```python
class Car:
    def __init__(self, make,model,year):
        self.make=make
        self.model=model
        self.year=year
        self.odometer=0
    
    def get_descriptive_name(self):
        print(f" The car is {self.make} and it's model is{self.model} and year is {self.year}")

    def get_your_car(self, seats):
        print(f"your car is {self.make}, {self.model} and number of seats are {seats}")

    def get_miles_read(self):
        print(f"Das Auto hat {self.odometer} miles fahren")
```

```python
my_car=Car("bmw","m8","2000")
my_car.get_descriptive_name()
my_car.get_your_car(4)
print(my_car.make)
print(my_car.model)
```

```python
my_new_car=Car(make="Mercdes",model="2000",year=2000)

my_new_car.get_descriptive_name()
```
