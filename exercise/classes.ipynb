```python
class Car:
    def __init__(self, make,model,year):
        self.make=make
        self.model=model
        self.year=year
        self.odometer=0
    
    def get_descriptive_name(self):
        print(f" The car is {self.make} and it's model is{self.model} and year is {self.year}")

    def get_your_car(self, seats):
        print(f"your car is {self.make}, {self.model} and number of seats are {seats}")

    def get_miles_read(self):
        print(f"Das Auto hat {self.odometer} miles fahren")
    
    # setter method
    def set_odometer(self,odo):

        if odo>=self.odometer:
            self.odometer=odo
        else:
            print("Smallest value can't be assigned here")
    
    def increment_odometer(self,value):

        self.odometer+=value
    
    def decrement_odometer(self,value):
        self.odometer=self.odometer-value
```

```python
my_car=Car("bmw","m8","2000")
my_car.get_descriptive_name()
my_car.get_your_car(4)
print(my_car.make)
print(my_car.model)
```

# Inheritance 

```python
class Electric_car(Car):
    def __init__(self,make,model,year):
        super().__init__(make,model,year)
        self.battery_size=32.6
        self.range=50

    def get_electic_data(self):
        print(f"The car {self.make} is electric")

    def get_battery_data(self):
        print(f"The batter size of {self.make} is {self.battery_size} V")
```

```python
new_electric=Electric_car("BMW","M11","2025")
new_electric.get_electic_data()
```

```python
new_electric.get_battery_data()
```

```python
new_electric.get_miles_read()
new_electric.set_odometer(100)
new_electric.increment_odometer(1000)
new_electric.get_miles_read()
```

# Polymorhism

## change the behaviour of the method of parent class in the child class by redefining it again and give it another functionality 

```python
class Mammal:

    def __init__(self,animal_type,legs):
        self.type=animal_type
        self.legs=legs
    
    def get_animal_description(self):
        print(f"The Animal is {self.animal_type} and it has {self.legs} legs!")
    
    def get_name(self):
        print(f" The name of Mammal is {self.type}")
```

```python
class Sparrow(Mammal):
    def __init__(self,animal_type,legs):
        super().__init__(animal_type,legs)

    def get_name(self):
        print(f" The sparrow has it's own type not mammal so it has 2 wings ")
```

```python
spa=Sparrow("reptile","4")
spa.get_name()
```
