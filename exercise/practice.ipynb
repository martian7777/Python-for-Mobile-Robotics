We want to create a function that validates whether an email is valid or not. Let's call this function validate_email_password(value, func). The function should have two parameters: a value and a function.

You first call the function passing as an argument an email and its corresponding  function that checks whether the email contains @ symbol

You call it a second time passing as an argument a password and its corresponding function that checks whether the length of the password is greater than 10.



```python
def check_email(email):
    if "@" in email:
        print ("Email is valid")
    else:
        print("Email is not valid")


def check_password(password):
    if len(password)>10:
        print("valid password")
    else:
        print("not valid password")

def validate_email_password(value,func):
    func(value)

    
validate_email_password("saqin@22",check_email)
validate_email_password("sferfre2233223",check_password)
```

## Exercise # 04

Since the two helping function of the above exercise will only be used in the  validate_email_password insert the checks as inner function to the outer function validate_email_password. If one of the enteries isn't correct print please change the  wrong entry. If both of them are not correct print please change both the enteries.



```python
def check_email_password_inner(email,password):

    #this is the inner function
    def check_email(email):
        return "@" in email
    #another inner function
    def check_password(password):
        return len(password)>10

    
    email_valid = check_email(email)
    password_valid = check_password(password)

    # Evaluate the results
    if not email_valid and not password_valid:
        print("Please change both entries")

    elif not email_valid:
        print("Please change the email")

    elif not password_valid:
        print("Please change the password")

    else:
        print("Both entries are correct")
```

```python
check_email_password_inner("genc@gmail.com", "mypassword123")

check_email_password_inner("gencgmail.com", "mypassword123")

check_email_password_inner("genc@gmail.com", "hello")

check_email_password_inner("gencgmail.com", "hello")
```
