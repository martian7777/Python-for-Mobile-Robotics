```python
def div(a,b):
    print(a/b)
        

def smart_div(func):

    def inner(a,b):
        if a<b:
            a,b=b,a
        
        return func(a,b)
    
    return inner
    

div=smart_div(div)
```

```python
div(2,4)
```

```python
import time
```

```python
def timer(func):

    def wrapper(*a):
        start=time.time()

        result= func(*a)
        end=time.time()
        print(f"{func.__name__}total time the function use",end-start)
        return result
    return wrapper
```

```python
@timer
def example_function(n):
    time.sleep(n)

example_function(2)
```

```python
# create a decorator to print the function name and
#  the values of it's argument every time function is called
```

```python
def print_function(func):

    def wrapper(*a):
            if a[:]>=(9,0):
             return func(a)
    
    return wrapper
```

```python

@print_function
def data_values(a):
    print(a)

```

```python
data_values(9,10,120)
```
