# Multiple Funtion Call 

```python
import numpy as np

def array_sum(data:np.array):
    return np.sum(data)
```

```python
list1=[1,2,34,55,65]
list2=[12,34,45,231]

print("Sum of list1 is ", array_sum(list1))
print("Sum of list2 is ", array_sum(list2))
```
