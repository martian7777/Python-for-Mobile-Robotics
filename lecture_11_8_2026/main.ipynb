```python
#keyword arguments

#** can be used as dic to store the result 

def user_build_profile(first, second, **user_info):
    profile={}
    profile['first name']=first
    profile['second name']=second
    for i,x in user_info.items():
        profile[i]=x

    print(profile)

user_build_profile("alex","rata",location="MIT",subject=122)
```

```python
def calculate_price(tax_rate):

    def get_tax(price):
        return price*tax_rate

    def get_price(price):
        return price +get_tax(price)
    
    return get_price


country1=calculate_price(12)

print(country1(100))
```

```python
def my_decorator(func_call):
    
    def wrapper():
        print("hello")
        func_call()
        print("lol")
    
    return wrapper

@my_decorator
def say_lol():
    print("lmao ")


say_lol()
```
